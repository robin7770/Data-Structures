# 1️⃣ Introduction to Frequency Dictionary

## 🔹 Definition

A **Frequency Dictionary** is a hashing-based technique used to count occurrences of elements in a dataset.

It uses:

```python
Dictionary (dict)
```

Where:

* Key → Element
* Value → Frequency (count)

---

# 2️⃣ Why Use Dictionary Instead of Frequency Array?

Use dictionary when:

✔ Element range is large
✔ Elements can be negative
✔ Elements are strings
✔ Values are unknown in advance

Example:

```python
arr = [1000000, -45, 9999999, 1000000]
```

You cannot create array of size 10 million easily.

So use:

```python
dict
```

# 3️⃣ Problem Statement

Given:

```python
n = [5,3,2,2,1,5,5,7,5,10]
m = [15,111,1,9,5,67,23]
```

For each element in `m`, print its frequency in `n`.

---

# 4️⃣ Step-by-Step Implementation

---

## Step 1: Create Empty Dictionary

```python
freq = {}
```

Initial state:

```
{}
```

---

## Step 2: Count Frequencies

```python
for num in n:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
```

After processing:

```
{
 1: 1,
 2: 2,
 3: 1,
 5: 4,
 7: 1,
 10: 1
}
```

---

## Cleaner Version (Recommended)

```python
for num in n:
    freq[num] = freq.get(num, 0) + 1
```

Explanation:

```
freq.get(num, 0)
```

If key exists → return value
If not → return 0

---

## Step 3: Process Query Array

```python
for num in m:
    print(freq.get(num, 0))
```

If element not found → returns 0








````python
# 5️⃣ Complete Code (Notebook Cell)

# -----------------------------
# Frequency Dictionary Example
# -----------------------------

n = [5,3,2,2,1,5,5,7,5,10]
m = [15,111,1,9,5,67,23]

# Step 1: Create dictionary
freq = {}

# Step 2: Count frequencies
for num in n:
    freq[num] = freq.get(num, 0) + 1

# Step 3: Process queries
print("Query Results:")
for num in m:
    print(f"{num} -> {freq.get(num, 0)}")
````

---

# 7️⃣ Complexity Analysis

## Time Complexity

Counting: O(n)
Query: O(m)

Total:

```
O(n + m)
```

Average case: O(1) per lookup

Worst case (rare): O(n)

---

## Space Complexity

In worst case:

```
O(n)
```

If all elements are unique.

# 1️⃣1️⃣ Example: Check Anagram Using Frequency Dict

```python
def is_anagram(s, t):
    if len(s) != len(t):
        return False
    
    freq = {}
    
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    
    for ch in t:
        if ch not in freq or freq[ch] == 0:
            return False
        freq[ch] -= 1
    
    return True

print(is_anagram("listen", "silent"))
```

Output:

```
True
```

# 1️⃣2️⃣ Example: Find First Repeating Element

```python
def first_repeating(arr):
    seen = {}
    
    for num in arr:
        if num in seen:
            return num
        seen[num] = 1
    
    return -1

print(first_repeating([1,2,3,2,4]))
```

Output:

```
2
```
