#  Selection Sort 

## 1️⃣ What is Selection Sort?

**Selection Sort** is a sorting algorithm that repeatedly:

1. Finds the **smallest element** in the unsorted portion of the array.
2. Places it at the **correct position** in the sorted portion.

The array is divided into two parts:

```
Sorted Part | Unsorted Part
```

Initially:

```
Sorted Part = empty
Unsorted Part = entire array
```

After every pass, **one element moves from unsorted to sorted part**.

---

# 2️⃣ Example Array

```
arr = [64, 25, 12, 22, 11]
```

Index positions:

```
0   1   2   3   4
64  25  12  22  11
```

---

````python
n = int(input())

arr = list(map(int,input().split()))

for i in range(n):

    min_index = i

    for j in range(i+1,n):

        if arr[j] < arr[min_index]:
            min_index = j

    arr[i], arr[min_index] = arr[min_index], arr[i]

print(*arr)
````

# Bubble Sort 

## What is Bubble Sort?

**Bubble Sort** repeatedly compares **adjacent elements** and swaps them if they are in the wrong order.

The largest element **“bubbles” to the end** after every pass.

Example:

```id="pbdp0r"
[5, 3, 8, 4, 2]
```

After first pass → largest number moves to the end.

---

# Example Array

```id="n4znl5"
arr = [5, 3, 8, 4, 2]
```

Indexes:

```id="dvyqpq"
0  1  2  3  4
5  3  8  4  2
```

---

````python
def bubbleSort(arr):
    n = len(arr)

    for i in range(n-2, -1, -1):
        swapped = False
        for j in range(0, i+1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr

n = int(input())
arr = list(map(int, input().split()))
print(*bubbleSort(arr))
````


---

# 1️⃣ Concept of Bubble Sort

Bubble Sort repeatedly **compares adjacent elements** and swaps them if they are in the wrong order.

After each pass:

* The **largest element moves to the end of the array**.

This process continues until the array becomes sorted.

Example:

```
Input  : [5,3,8,4,2]
Output : [2,3,4,5,8]
```

The largest element **"bubbles up" to the end**, like a bubble rising in water.

# Insertion Sort 
---

# 1️⃣ What is Insertion Sort?

Insertion Sort is a **simple sorting algorithm** that builds the sorted array **one element at a time**. ([GeeksforGeeks][1])

Idea:

```text
Take one element
Insert it in the correct position
inside the sorted part
```

---

# 2️⃣ Real-Life Example (Playing Cards)

Imagine sorting **cards in your hand**.

Example:

```text
Cards in hand
7 4 5 2
```

Steps:

1. Start with first card → already sorted
2. Pick next card
3. Insert it into correct position

Exactly the same thing happens in **Insertion Sort**. ([Code And Debug][2])

---

# 3️⃣ Main Idea

Array is divided into **two parts**

```text
[ Sorted | Unsorted ]
```

Example:

```text
[4 | 3 1 2]
```

* Left side → sorted
* Right side → unsorted

We take one element from the **unsorted side** and insert it into the correct place in the **sorted side**.

---

# 4️⃣ Algorithm Steps

1️⃣ Assume first element is already sorted

2️⃣ Start from second element

3️⃣ Compare it with previous elements

4️⃣ Shift larger elements to the right

5️⃣ Insert the element in correct position

6️⃣ Repeat until array is sorted

---


````python
def insertionSort(arr):

    n = len(arr)

    for i in range(1,n):

        key = arr[i]

        j = i-1

        while j >= 0 and arr[j] > key:

            arr[j+1] = arr[j]

            j -= 1

        arr[j+1] = key

    return arr


n = int(input())
arr = list(map(int,input().split()))

print(*insertionSort(arr))
````

# Merge Sort Algorithm 

---

# 1. Introduction

Merge Sort is a **Divide and Conquer sorting algorithm**.

It works in three main steps:

```
Divide → Conquer → Combine
```

1. **Divide** – Split the array into two halves
2. **Conquer** – Recursively sort each half
3. **Combine** – Merge the sorted halves together

Merge sort guarantees **O(n log n)** time complexity in **best, average, and worst cases**, which makes it efficient for large datasets. ([Code and Debug][1])

---

# 2. How Merge Sort Works

Example array:

```
[38, 27, 43, 10]
```

### Step 1 — Divide

```
[38,27,43,10]

→

[38,27]     [43,10]

→

[38] [27] [43] [10]
```

Each element becomes a **single element array**, which is already sorted.

---

### Step 2 — Merge

Now merge the sorted pieces.

```
[38] [27] → [27,38]

[43] [10] → [10,43]

[27,38] + [10,43]

→ [10,27,38,43]
```

Final sorted array.

---

# 3. The `mergeSort` Function

````python
def mergeSort(self, arr, l, r):

    if l < r:

        mid = (l + r) // 2

        # Sort first and second halves
        self.mergeSort(arr, l, mid)
        self.mergeSort(arr, mid + 1, r)

        self.merge(arr, l, mid, r)
````

---

# 4. Explanation of `mergeSort`

### Base Condition

```
if l < r
```

If the array has **0 or 1 element**, it is already sorted.

Example

```
[5]
```

No sorting needed.

---

### Finding the Midpoint

```
mid = (l + r) // 2
```

Example

```
l = 0
r = 3

mid = 1
```

Split array:

```
left  = arr[0..1]
right = arr[2..3]
```

---

### Recursive Calls

```
self.mergeSort(arr, l, mid)
self.mergeSort(arr, mid+1, r)
```

This recursively divides the array until each subarray has **one element**.

---



# 5. The `merge` Function

````python
def merge(self, arr, l, mid, r):

    # Create temp arrays
    left = arr[l:mid+1]
    right = arr[mid+1:r+1]

    i = 0
    j = 0
    k = l

    # Merge the temp arrays back
    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            arr[k] = left[i]
            i += 1

        else:
            arr[k] = right[j]
            j += 1

        k += 1

    # Copy remaining elements
    while i < len(left):
        arr[k] = left[i]
        i += 1
        k += 1

    while j < len(right):
        arr[k] = right[j]
        j += 1
        k += 1
````


# 6. Explanation of Merge Function

### Step 1 — Create Temporary Arrays

```
left = arr[l:mid+1]
right = arr[mid+1:r+1]
```

Example

```
left  = [38,27]
right = [43,10]
```

---

### Step 2 — Initialize Pointers

```
i = 0
j = 0
k = l
```

| Pointer | Purpose              |
| ------- | -------------------- |
| i       | index of left array  |
| j       | index of right array |
| k       | index of main array  |

---

### Step 3 — Compare Elements

```
while i < len(left) and j < len(right)
```

Compare

```
left[i] vs right[j]
```

Smallest element goes into main array.

Example

```
27 vs 10 → 10
```

---

### Step 4 — Copy Remaining Elements

If one array finishes first, copy the rest.

Example

```
left  = [27,38]
right = [10,43]

Result

[10,27,38,43]
```

---

# 7. Complete Code

````python
class Solution:

    def mergeSort(self, arr, l, r):

        if l < r:

            mid = (l + r) // 2

            self.mergeSort(arr, l, mid)
            self.mergeSort(arr, mid + 1, r)

            self.merge(arr, l, mid, r)


    def merge(self, arr, l, mid, r):

        left = arr[l:mid+1]
        right = arr[mid+1:r+1]

        i = 0
        j = 0
        k = l

        while i < len(left) and j < len(right):

            if left[i] <= right[j]:

                arr[k] = left[i]
                i += 1

            else:

                arr[k] = right[j]
                j += 1

            k += 1


        while i < len(left):

            arr[k] = left[i]
            i += 1
            k += 1


        while j < len(right):

            arr[k] = right[j]
            j += 1
            k += 1
````

````python
n = int(input())                     # size of array
arr = list(map(int, input().split()))  # elements

solution = Solution()
solution.mergeSort(arr, 0, n-1)

print(*arr)
````

# Quick Sort in Python

---

# 1. Introduction

Quick Sort is a **Divide and Conquer sorting algorithm**.

Steps of Quick Sort:

```
1. Choose a Pivot
2. Partition the Array
3. Recursively Sort Left and Right
```

* A **pivot element** is selected.
* Elements **≤ pivot go to the left**.
* Elements **> pivot go to the right**.
* The process repeats recursively. ([codeanddebug.in][1])

---

# 2. Example

Array:

```

```

# 3. Partition Function (Code from Blog)

The **partition function** arranges elements around the pivot.

### Code

````python
def partition(arr, low, high):

    pivot = arr[low]

    i = low
    j = high

    while i < j:

        while i <= high - 1 and arr[i] <= pivot:
            i += 1

        while j >= low + 1 and arr[j] > pivot:
            j -= 1

        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    arr[low], arr[j] = arr[j], arr[low]

    return j
````

---

# 4. Explanation of Partition

### Pivot Selection

```
pivot = arr[low]
```

The **first element** is chosen as pivot.

Example

```
[6,3,9,5,2,8]
pivot = 6
```

---

### Two Pointers

```
i = low
j = high
```

| Pointer | Purpose          |
| ------- | ---------------- |
| i       | moves from left  |
| j       | moves from right |

---

### Move `i` Forward

```
while arr[i] <= pivot
```

Move until a value **greater than pivot** is found.

---

### Move `j` Backward

```
while arr[j] > pivot
```

Move until a value **smaller than pivot** is found.

---

### Swap Elements

```
arr[i], arr[j] = arr[j], arr[i]
```

This places:

* smaller elements on left
* larger elements on right.

---

### Final Pivot Placement

```
arr[low], arr[j] = arr[j], arr[low]
```

Pivot goes to its **correct sorted position**.

# 5. Quick Sort Function

### Code

````python

def quick_sort(arr, low, high):

    if low < high:

        pIndex = partition(arr, low, high)

        quick_sort(arr, low, pIndex - 1)

        quick_sort(arr, pIndex + 1, high)

````

# 7. Complete Code (Same Logic)

````python

def partition(arr, low, high):

    pivot = arr[low]

    i = low
    j = high

    while i < j:

        while i <= high-1 and arr[i] <= pivot:
            i += 1

        while j >= low+1 and arr[j] > pivot:
            j -= 1

        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    arr[low], arr[j] = arr[j], arr[low]

    return j


def quick_sort(arr, low, high):

    if low < high:

        pIndex = partition(arr, low, high)

        quick_sort(arr, low, pIndex-1)
        quick_sort(arr, pIndex+1, high)
````
