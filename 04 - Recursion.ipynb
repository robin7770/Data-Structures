# 1️⃣ What is Recursion?

**Recursion** is a programming technique where a **function calls itself** to solve a smaller version of the same problem.

Instead of using loops, the function **repeats itself** until a stopping condition is reached.

General form:

```python
def func():
    if base_condition:
        return
    func()
```

---

# 2️⃣ Two Important Parts of Recursion

### 1️⃣ Base Case

The condition where recursion **stops**.

Example:

```python
if n == 0:
    return
```

### 2️⃣ Recursive Case

The function calls itself.

```python
func(n-1)
```

---

# 3️⃣ How Recursion Works (Call Stack)

Example:

```
factorial(4)
```

Execution:

```
factorial(4)
→ 4 * factorial(3)

factorial(3)
→ 3 * factorial(2)

factorial(2)
→ 2 * factorial(1)

factorial(1)
→ 1
```

Then stack resolves:

```
1
2 × 1 = 2
3 × 2 = 6
4 × 6 = 24
```

Final result:

```
24
```

---

# 4️⃣ Types of Recursion

### 1️⃣ Direct Recursion

Function calls itself.

```
f() → f()
```

### 2️⃣ Indirect Recursion

```
f() → g()
g() → f()
```

### 3️⃣ Tail Recursion

Recursive call is the **last statement**.

### 4️⃣ Head Recursion

Recursive call happens **first**.

| Code size | Smaller        | Longer    |

# 6️⃣ Factorial Using Recursion

Factorial formula:

[
n! = n \times (n-1)!
]

Base case:

[
0! = 1
]

---






````python

def factorial(n):
    
    # Base case
    if n == 0 or n == 1:
        return 1
    
    # Recursive case
    return n * factorial(n-1)


print(factorial(5))
````

# 7️⃣ Reverse Array Using Recursion

Example:

```
arr = [1,2,3,4,5]
```

Output:

```
[5,4,3,2,1]
```

---

# Method 1: Using Two Pointers

Swap first and last elements recursively.

---

````python
def reverse_array(arr, left, right):
    
    if left >= right:
        return
    
    # swap
    arr[left], arr[right] = arr[right], arr[left]
    
    # recursive call
    reverse_array(arr, left+1, right-1)


arr = [1,2,3,4,5]

reverse_array(arr, 0, len(arr)-1)

print(arr)
````



---

# Step-by-Step Execution

Initial array:

```
[1,2,3,4,5]
```

Step 1:

```
swap 1 and 5
[5,2,3,4,1]
```

Step 2:

```
swap 2 and 4
[5,4,3,2,1]
```

Stop when:

```
left >= right
```

# 8️⃣ Reverse Array Using One Pointer

Another recursion method.

---

````python
def reverse_array(arr, i, n):
    
    if i >= n//2:
        return
    
    arr[i], arr[n-i-1] = arr[n-i-1], arr[i]
    
    reverse_array(arr, i+1, n)


arr = [1,2,3,4,5]

reverse_array(arr, 0, len(arr))

print(arr)
````

# 1️⃣2️⃣ Example – Sum of N Numbers

Formula:

```
sum(n) = n + sum(n-1)
```

Code:

````python
def sum_n(n):
    
    if n == 0:
        return 0
    
    return n + sum_n(n-1)

print(sum_n(5))
````


---

# 1️⃣3️⃣ Example – Print 1 to N

````python
def print_numbers(n):
    
    if n == 0:
        return
    
    print_numbers(n-1)
    
    print(n)

print_numbers(5)
````
