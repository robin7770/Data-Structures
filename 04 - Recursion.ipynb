# 1️⃣ What is Recursion?

**Recursion** is a programming technique where a **function calls itself** to solve a smaller version of the same problem.

Instead of using loops, the function **repeats itself** until a stopping condition is reached.

General form:

```python
def func():
    if base_condition:
        return
    func()
```

---

# 2️⃣ Two Important Parts of Recursion

### 1️⃣ Base Case

The condition where recursion **stops**.

Example:

```python
if n == 0:
    return
```

### 2️⃣ Recursive Case

The function calls itself.

```python
func(n-1)
```

---

# 3️⃣ How Recursion Works (Call Stack)

Example:

```
factorial(4)
```

Execution:

```
factorial(4)
→ 4 * factorial(3)

factorial(3)
→ 3 * factorial(2)

factorial(2)
→ 2 * factorial(1)

factorial(1)
→ 1
```

Then stack resolves:

```
1
2 × 1 = 2
3 × 2 = 6
4 × 6 = 24
```

Final result:

```
24
```

---

# 4️⃣ Types of Recursion

### 1️⃣ Direct Recursion

Function calls itself.

```
f() → f()
```

### 2️⃣ Indirect Recursion

```
f() → g()
g() → f()
```

### 3️⃣ Tail Recursion

Recursive call is the **last statement**.

### 4️⃣ Head Recursion

Recursive call happens **first**.

---

# 6️⃣ Factorial Using Recursion

Factorial formula:

[
n! = n \times (n-1)!
]

Base case:

[
0! = 1
]

---






````python

def factorial(n):
    
    # Base case
    if n == 0 or n == 1:
        return 1
    
    # Recursive case
    return n * factorial(n-1)


print(factorial(5))
````

# 7️⃣ Reverse Array Using Recursion

Example:

```
arr = [1,2,3,4,5]
```

Output:

```
[5,4,3,2,1]
```

---

# Method 1: Using Two Pointers

Swap first and last elements recursively.

---

````python
def reverse_array(arr, left, right):
    
    if left >= right:
        return
    
    # swap
    arr[left], arr[right] = arr[right], arr[left]
    
    # recursive call
    reverse_array(arr, left+1, right-1)


arr = [1,2,3,4,5]

reverse_array(arr, 0, len(arr)-1)

print(arr)
````



---

# Step-by-Step Execution

Initial array:

```
[1,2,3,4,5]
```

Step 1:

```
swap 1 and 5
[5,2,3,4,1]
```

Step 2:

```
swap 2 and 4
[5,4,3,2,1]
```

Stop when:

```
left >= right
```

# 8️⃣ Reverse Array Using One Pointer

Another recursion method.

---

````python
def reverse_array(arr, i, n):
    
    if i >= n//2:
        return
    
    arr[i], arr[n-i-1] = arr[n-i-1], arr[i]
    
    reverse_array(arr, i+1, n)


arr = [1,2,3,4,5]

reverse_array(arr, 0, len(arr))

print(arr)
````

# 1️⃣2️⃣ Example – Sum of N Numbers

Formula:

```
sum(n) = n + sum(n-1)
```

Code:

````python
def sum_n(n):
    
    if n == 0:
        return 0
    
    return n + sum_n(n-1)

print(sum_n(5))
````

# 1️⃣3️⃣ Example – Print 1 to N

````python
def print_numbers(n):
    
    if n == 0:
        return
    
    print_numbers(n-1)
    
    print(n)

print_numbers(5)
````

#  What is a Palindrome?

A **Palindrome** is a string that reads the **same forward and backward**.

Examples:

```
madam
racecar
level
```

Not palindrome:

```
hello
world
```

---

# 🧠 Idea Using Recursion

We compare:

```
first character  ↔ last character
```

If they match → move inward.

Example:

```
madam
m == m ✔
a == a ✔
d == d ✔
```

Stop when pointers cross.

---

# ⚙️ Recursive Logic

1️⃣ Base Case

```
if left >= right
return True
```

Meaning we checked all characters.

2️⃣ If characters not equal

```
return False
```

3️⃣ Otherwise move inward

```
check(left+1 , right-1)
```

---

````python
def is_palindrome(s, left, right):
    
    # Base condition
    if left >= right:
        return True
    
    # If characters don't match
    if s[left] != s[right]:
        return False
    
    # Recursive call
    return is_palindrome(s, left+1, right-1)


# Test
s = "madam"

print(is_palindrome(s, 0, len(s)-1))
````

# 🚀 Cleaner Version (Best for Interviews)

```python
def palindrome(s, i=0):
    
    if i >= len(s)//2:
        return True
    
    if s[i] != s[len(s)-i-1]:
        return False
    
    return palindrome(s, i+1)


s = "level"

print(palindrome(s))
```

Output

```
True
```
