# 1️⃣ Introduction to Hashing

## 🔹 Definition

**Hashing** is a technique used to store and retrieve data efficiently by mapping keys to specific indices using a hash function.

[
\text{Index} = h(\text{key})
]

It allows:

* Fast lookup
* Fast insertion
* Fast deletion

Average Time Complexity: **O(1)**

---

````python
# Frequency array for n
freq = [0] * 11   # index 0 unused

n = [5,3,2,2,1,5,5,7,5,10]
m = [15,111,1,9,5,67,23]

# Count frequency of n
for num in n:
    freq[num] += 1

# Check elements of m
for num in m:
    if 1 > num or num > 10:
        print(0)
    else:
        print(freq[num])
````

# 2️⃣ What is Direct Addressing?

## 🔹 Definition

Direct Addressing is a special case of hashing where:

[
h(x) = x
]

That means:

* The value itself acts as an index.
* No separate hash function required.
* No collision occurs (if range is small).

---

# 3️⃣ When to Use Frequency Array?

Use Frequency Array when:

✔ Element range is small
✔ Data size is very large
✔ Need fast frequency queries
✔ Constraints like:

```
1 ≤ arr[i] ≤ 10
```

# 6️⃣ Complete Code (Notebook Cell)

# -----------------------------

# -----------------------------

# Input arrays
n = [5,3,2,2,1,5,5,7,5,10]
m = [15,111,1,9,5,67,23]

# Step 1: Create frequency array
freq = [0] * 11

# Step 2: Count frequency of n
for num in n:
    freq[num] += 1

# Step 3: Process queries
print("Query Results:")
for num in m:
    if num < 1 or num > 10:
        print(f"{num} -> 0")
    else:
        print(f"{num} -> {freq[num]}")
```

# 8️⃣ Complexity Analysis

## Time Complexity

Counting: O(n)
Query Processing: O(m)

Total:

[
O(n + m)
]

Works even if:

```
n = 10^8
m = 10^8
```

---

## Space Complexity

Frequency array size = 11

[
O(1)
]

Constant space.

---


````python
# using dictionary
freq = {}

for num in n:
    freq[num] = freq.get(num, 0) + 1

for num in m:
    print(freq.get(num, 0))
````

# 1️⃣2️⃣ Counting Sort Using Same Concept

```python
def counting_sort(arr):
    max_val = max(arr)
    freq = [0] * (max_val + 1)

    for num in arr:
        freq[num] += 1

    sorted_arr = []
    for i in range(len(freq)):
        sorted_arr.extend([i] * freq[i])

    return sorted_arr

print(counting_sort([5,3,2,2,1,5]))
```
